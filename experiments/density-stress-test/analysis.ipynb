```python
import numpy as np
import matplotlib.pyplot as plt
import os 
import sys
sys.path.append(os.path.expanduser("~")+"/src/at-gpflow/")
print(sys.path)
```

```python
data = np.load("/home/janneke/src/at-gpflow/experiments/density-stress-test/results-times-fixed-target.npz")
print(dict(data).keys())
data_full_gp_fixed = data["full"].mean(axis=1)
data_sparse_gp_fixed = data["sparse"].mean(axis=1)
data_svgp_fixed = data["svgp_times"].mean(axis=1)
plt.rcParams["font.family"] = "serif"
fig, ax = plt.subplots(1, 2, sharey=True, figsize=(12, 5))

X = np.logspace(6, 13, 10, base=2)[:-1]
fig.supylabel("$\log_2$ time (s)", fontsize=20)
ax[0].set_xlabel("$\log_2$ # source points", fontsize=20)
ax[0].plot(np.log2(X), np.log2(data_svgp_fixed), color="red", lw=3, marker="|", ms=12, mew=3, label="full CMOGP")
ax[0].plot(np.log2(X), np.log2(data_full_gp_fixed), color="darkcyan", lw=3, marker="|", ms=12, mew=3, label="full CMOGP")
ax[0].plot(np.log2(X), np.log2(data_sparse_gp_fixed), color="darkviolet", lw=3, marker="|", ms=12, mew=3, label="sparse CMOGP (ours)")
ax[0].legend(fontsize=13)


data = np.load("/home/janneke/src/at-gpflow/experiments/density-stress-test/results-times-variable-target.npz")
data_full_gp = data["full"].mean(axis=1)
data_sparse_gp = data["sparse"].mean(axis=1)
ax[0].tick_params(axis="both", labelsize=15)
ax[1].tick_params(axis="both", labelsize=15)
X = np.logspace(6, 13, 10, base=2)[:-1]
ax[1].set_xlabel("$\log_2$ #source points", fontsize=20)
ax[1].plot(np.log2(X), np.log2(data_full_gp), color="darkgreen", lw=3, marker="|", ms=12, mew=3, label="full CMOGP")
ax[1].plot(np.log2(X), np.log2(data_sparse_gp), color="darkmagenta", lw=3, marker="|", ms=12, mew=3, label="sparse CMOGP (ours)")
ax[1].legend(fontsize=13)
plt.tight_layout()
plt.show()

```
