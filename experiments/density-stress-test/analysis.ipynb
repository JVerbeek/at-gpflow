```python
import numpy as np
import matplotlib.pyplot as plt
import os 
import sys
sys.path.append(os.path.expanduser("~")+"/src/at-gpflow/")
print(sys.path)
```

```python
data_cmogp = np.load("/home/janneke/src/at-gpflow/experiments/density-stress-test/results-CMOGP-variable-target.npz")
data_scmogp = np.load("/home/janneke/src/at-gpflow/experiments/density-stress-test/results-sCMOGP-variable-target.npz")
data_svgp = np.load("/home/janneke/src/at-gpflow/experiments/density-stress-test/results-SVGP-variable-target.npz")

times_cmogp = data_cmogp["times"].mean(axis=1)
times_scmogp = data_scmogp["times"].mean(axis=1)
print(times_scmogp)
print(times_cmogp)
plt.rcParams["font.family"] = "serif"
fig, ax = plt.subplots(1, 2, sharey=True, figsize=(12, 5))
X = np.logspace(6, 14, 8, base=2)
fig.supylabel("$\log_2$ time (s)", fontsize=20)
ax[0].set_xlabel("$\log_2$ # source points", fontsize=20)
ax[0].plot(np.log2(X), np.log2(times_cmogp), color="red", lw=3, marker="|", ms=12, mew=3, label="full CMOGP")
ax[0].plot(np.log2(X), np.log2(times_scmogp), color="darkcyan", lw=3, marker="|", ms=12, mew=3, label="full sCMOGP")
ax[0].legend(fontsize=13)


ax[0].tick_params(axis="both", labelsize=15)
ax[1].tick_params(axis="both", labelsize=15)
X = np.logspace(6, 13, 10, base=2)[:-1]
ax[1].set_xlabel("$\log_2$ #source points", fontsize=20)
ax[1].legend(fontsize=13)
plt.tight_layout()
plt.show()

```
