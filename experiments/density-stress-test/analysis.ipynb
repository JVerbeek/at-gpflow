```python
import numpy as np
import matplotlib.pyplot as plt
import os 
import sys
sys.path.append(os.path.expanduser("~")+"/src/at-gpflow/")
print(sys.path)
```

```python
data = np.load("/home/janneke/src/at-gpflow/experiments/density-stress-test/results/times-with-ivs.npz")
data_full_gp = data["full_times"].mean(axis=1)[:-1]
data_sparse_gp = data["sparse_times"].mean(axis=1)[:-1]

X = np.logspace(6, 13, 10, base=2)[:-1]
plt.figure(figsize=(12, 3))
plt.ylabel("$\log_2$ time (s)", fontsize=20)
plt.xlabel("$\log_2$ # source points", fontsize=20)
plt.plot(np.log2(X), np.log2(data_full_gp), color="darkgreen", lw=3, marker="|", ms=12, mew=3, label="full CMOGP")
plt.plot(np.log2(X), np.log2(data_sparse_gp), color="darkmagenta", lw=3, marker="|", ms=12, mew=3, label="sparse CMOGP (ours)")
plt.legend(fontsize=13)
plt.show()


X = np.logspace(6, 13, 10, base=2)[:-1]
plt.ylabel("$\log_2$ time (s)")
plt.xlabel("log_2 # source points")
plt.plot(X, data_full_gp)
plt.plot(X, data_sparse_gp)

data = np.load("/home/janneke/src/at-gpflow/experiments/density-stress-test/results/times.npz")
data_full_gp = data["full_times"].mean(axis=1)[:-1]
data_sparse_gp = data["sparse_times"].mean(axis=1)[:-1]

X = np.logspace(6, 13, 10, base=2)[:-1]
plt.figure(figsize=(12, 3))
plt.ylabel("$\log_2$ time (s)", fontsize=20)
plt.xlabel("$\log_2$ # source points", fontsize=20)
plt.plot(np.log2(X), np.log2(data_full_gp), lw=3, marker="|", ms=12, mew=3, label="full C-MOGP")
plt.plot(np.log2(X), np.log2(data_sparse_gp), lw=3, marker="|", ms=12, mew=3, label="sparse C-MOGP (ours)")
plt.legend(fontsize=13)
plt.show()


X = np.logspace(6, 13, 10, base=2)[:-1]
plt.ylabel("$\log_2$ time (s)")
plt.xlabel("log_2 # source points")
plt.plot(X, data_full_gp)
plt.plot(X, data_sparse_gp)
```
