```python
import numpy as np
import matplotlib.pyplot as plt
import os 
import sys
from matplotlib import color_sequences
sys.path.append(os.path.expanduser("~")+"/src/at-gpflow/")
print(sys.path)
```

```python

fig, (ax1, ax2) = plt.subplots(1, 2, sharey=True, figsize=(12, 5))
for ax, mode in zip((ax1, ax2), ("fixed", "variable")):
    X = np.logspace(6, 16, 10, base=2) 
    data_cmogp = np.load(f"/home/janneke/src/at-gpflow/experiments/density-stress-test/results-CMOGP-{mode}-target.npz")
    times_cmogp = data_cmogp["times"].mean(axis=1)
    print(times_cmogp)
    data_scmogp = np.load(f"/home/janneke/src/at-gpflow/experiments/density-stress-test/results-sCMOGP-{mode}-target.npz")
    data_svgp = np.load(f"/home/janneke/src/at-gpflow/experiments/density-stress-test/results-SVGP-{mode}-target.npz")
    times_scmogp = data_scmogp["times"].mean(axis=1)
    print(times_scmogp)
    times_svgp = data_svgp["times"].mean(axis=1)
    print(times_svgp)
    plt.rcParams["font.family"] = "serif"
    fig.supylabel("log$_2$ time (s)", fontsize=20)
    colors = color_sequences["Set2"]
    print(ax)
    print(np.log2(times_cmogp))
    ax.set_xlabel("log$_2$ # source points", fontsize=20)
    ax.plot(np.log2(X), np.log2(times_svgp), color=colors[0], lw=3, marker="|", ms=12, mew=3, label="SVGP")
    ax.plot(np.log2(X), np.log2(times_cmogp), color=colors[1], lw=3, marker="|", ms=12, mew=3, label="full CMOGP")
    ax.plot(np.log2(X), np.log2(times_scmogp), color=colors[2], lw=3, marker="|", ms=12, mew=3, label="sCMOGP")
    ax.legend(fontsize=13)
    ax.set_title(f"{mode} target", fontsize=20)
    ax.tick_params(axis="both", labelsize=15)
    ax.tick_params(axis="both", labelsize=15)
    plt.tight_layout()
plt.show()

```
