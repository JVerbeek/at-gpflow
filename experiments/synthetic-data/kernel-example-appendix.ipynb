```python
from gpflow.kernels import RBF, Matern12, Matern32
import numpy as np
import matplotlib.pyplot as plt
import matplotlib as mpl
colors = mpl.color_sequences["Set1"]
plt.rcParams["font.family"] = "serif"

fig, ax = plt.subplots(4, 1, figsize=(16, 9))
X = np.linspace(-10, 10, 200)
names = ["RBF", "Matern12", "Matern32"]
klist = [RBF(), Matern12(), Matern32()]
for i, k in enumerate([RBF(), Matern12(), Matern32()]):
    for j in range(3):
        sample = np.random.multivariate_normal(np.zeros((len(X),)), k(X[:,None]))
        ax[i].plot(X, sample, color=colors[i])
        ax[i].set_title(names[i], fontsize=15)
        ax[i].tick_params(axis="both", labelsize=15)
for j in range(3):
    sample = np.random.multivariate_normal(np.zeros((len(X),)), (klist[1] + klist[2])((X[:,None])))
    ax[-1].plot(X, sample)
    ax[-1].set_title("Matern12 + Matern32", fontsize=15)
    ax[-1].tick_params(axis="both", labelsize=15)
plt.tight_layout()
plt.show()
```
