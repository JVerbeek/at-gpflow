```python
import numpy as np 
import matplotlib.pyplot as plt 
```

```python
res = np.load("/home/janneke/repos/at-gpflow/experiments/check_approximation_quality/results_sweep.npz")

cmogp_results = res["cmogp_mse"]
vgp_results = res["vgp_mse"]
```

```python
mean_res_cmogp = cmogp_results.mean(axis=1)
std_res_cmogp = cmogp_results.std(axis=1)
mean_res_vgp = vgp_results.mean(axis=1)
std_res_vgp = vgp_results.std(axis=1)

props = np.arange(0.05, 2, 0.05).round(3)
plt.plot(props, mean_res_cmogp, label="cmogp")
plt.plot(props, mean_res_vgp, label="vgp")
plt.fill_between(props, mean_res_cmogp + std_res_cmogp, mean_res_cmogp - std_res_cmogp, color="C0", alpha=0.2)
plt.fill_between(props, mean_res_vgp + std_res_vgp, mean_res_vgp - std_res_vgp, color="C1", alpha=0.2)
plt.xticks(props, props, rotation=90)
plt.xlabel("target size as proportion of source")
plt.ylabel("MSE")
plt.legend()
plt.show()
```
