```python
import matplotlib.pyplot as plt 
import numpy as np
import gpflow as gpf 
```

```python
X = np.linspace(0, 10, 100)
k = gpf.kernels.RBF()
f = np.random.multivariate_normal(np.zeros_like(X), k(X[:,None]))
y1 = f + np.random.normal(0, 0.1, 100)
X2 = [x for i, x in enumerate(X) if i % 10 == 0]
y2 = [y for i, y in enumerate(-f + np.random.normal(0, 0.2, 100)) if i % 10 == 0]
plt.figure(figsize=(12, 4))
plt.plot(X, f, linewidth=3, alpha=0.3)
plt.plot(X, -f, linewidth=3, alpha=0.3)
plt.plot(X, y1, color="C0", marker="x", lw=0, mew=2, ms=15)
plt.plot(X2, y2, color="C1", marker="x", lw=0, mew=3, ms=15)
plt.tick_params(
    axis='both',          # changes apply to the x-axis
    which='both',      # both major and minor ticks are affected
    labelleft=False,
    labelbottom=False
) # labels along the bottom edge are off
plt.xlabel("$x$", fontsize=25)
plt.ylabel("$y$", fontsize=25)
plt.show()

X = np.arange(0, 100, 1)
plt.figure(figsize=(6, 6))
plt.fill_between(X[:31], 70, 100, color="C1", alpha=0.3)
plt.fill_between(X[30:], 0, 70, color="C0", alpha=0.3)
plt.fill_between(X[30:], 70, 100, color="purple", alpha=0.3)
plt.fill_between(X[:31], 0, 70, color="purple", alpha=0.3)
```
