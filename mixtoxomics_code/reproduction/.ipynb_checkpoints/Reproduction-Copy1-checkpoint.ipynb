## Load the data

```python
imi_cypro_reproduction = pd.read_excel("imi_cypro.xlsx")
imi_clo_reproduction = pd.read_excel("imi_clo.xlsx")
thia_cypro_reproduction = pd.read_excel("thia_cypro.xlsx")
```

```python
imi_cypro_reproduction = imi_cypro_reproduction.dropna()
imi_clo_reproduction = imi_clo_reproduction.dropna()
thia_cypro_reproduction = thia_cypro_reproduction.dropna()
```

```python
## Imi Clo
```

```python
X_imi_clo = np.concatenate([imi_clo_reproduction['concentration (mg/kg) Imidacloprid'].to_numpy().reshape(-1,1), imi_clo_reproduction['concentration (mg/kg) Clothianidin'].to_numpy().reshape(-1,1)], axis=1)
Y_imi_clo = imi_clo_reproduction['juveniles'].to_numpy().reshape(-1,1)
Y_imi_clo = (Y_imi_clo - np.min(Y_imi_clo))/(np.max(Y_imi_clo)-np.min(Y_imi_clo))
```

```python
X_imi_cypro = np.concatenate([imi_cypro_reproduction['concentration (mg/kg) Imidacloprid'].to_numpy().reshape(-1,1), imi_cypro_reproduction['concentration (mg/kg) Cyproconazole'].to_numpy().reshape(-1,1)], axis=1)
Y_imi_cypro = imi_cypro_reproduction['juveniles'].to_numpy().reshape(-1,1)
Y_imi_cypro = (Y_imi_cypro-np.min(Y_imi_cypro))/(np.max(Y_imi_cypro)-np.min(Y_imi_cypro))
```

```python
X_thia_cypro = np.concatenate([thia_cypro_reproduction['Concentration Thiacloprid'].to_numpy().reshape(-1,1), thia_cypro_reproduction['Concentration Cyproconazole'].to_numpy().reshape(-1,1)], axis=1)
Y_thia_cypro = thia_cypro_reproduction['juveniles'].to_numpy().reshape(-1,1)
Y_thia_cypro = (Y_thia_cypro-np.min(Y_thia_cypro))/(np.max(Y_thia_cypro)-np.min(Y_thia_cypro))
```

```python
imi_clo_reproduction.columns = ['ToxicUnitsIMI', 'ToxicUnitsCLO', 'drug1.conc', 'drug2.conc',  'Adults', 'effect']
```

```python
imi_cypro_reproduction.columns = ['ToxicUnitsIMI', 'ToxicUnitsCYPRO', 'drug1.conc', 'drug2.conc',  'Adults', 'effect']
```

```python
thia_cypro_reproduction.columns = ['ToxicUnitsCYPRO', 'ToxicUnitsTHIA', 'drug1.conc', 'drug2.conc', 'solvent', 'Adults', 'effect']
```

```python
## build and fit two models
```

```python
data_imi_clo = (X_imi_clo, Y_imi_clo)

k = K_multiplicative()

m1 = gpflow.models.GPR(data_imi_clo, kernel=k, mean_function=None)

opt = gpflow.optimizers.Scipy()
opt_logs = opt.minimize(m1.training_loss, m1.trainable_variables, options=dict(maxiter=100))
```

```python
data_imi_cypro = (X_imi_cypro, Y_imi_cypro)

k = K_multiplicative()

m2 = gpflow.models.GPR(data_imi_cypro, kernel=k, mean_function=None)

opt = gpflow.optimizers.Scipy()
opt_logs = opt.minimize(m2.training_loss, m2.trainable_variables, options=dict(maxiter=100))
```

```python
data_thia_cypro = (X_thia_cypro, Y_thia_cypro)
#k1 = gpflow.kernels.RBF(active_dims=[0])
#k2 = gpflow.kernels.RBF(active_dims=[1])

#k = k1*k2

k = K_multiplicative()


m3 = gpflow.models.GPR(data_thia_cypro, kernel=k, mean_function=None)

#m1.kernel.kernels[0].lengthscales.assign(np.max(data_imi_clo[0][:,0]/3))
#m1.kernel.kernels[1].lengthscales.assign(np.max(data_imi_clo[0][:,1]/3))
#m1.kernel.kernels[0].variance.assign(np.max(data_imi_clo[1]/2))
#m1.kernel.kernels[1].variance.assign(np.max(data_imi_clo[1]/2))

m3.kernel.variance_da.assign(np.nanmax(data_thia_cypro[1]/3))
m3.kernel.lengthscale_da.assign(np.nanmax(data_thia_cypro[0][:,0]/2))
m3.kernel.lengthscale_db.assign(np.nanmax(data_thia_cypro[0][:,1]/2))

opt = gpflow.optimizers.Scipy()
opt_logs = opt.minimize(m3.training_loss, m3.trainable_variables, options=dict(maxiter=100))
```
