```python
def compute_mse(X_replicates, X, Y, m):
    # Predict the mean and covariance of the GP fit at the test locations
    mean2, Cov2 = m.predict_f(X_replicates)

    mean_shape = mean2.shape[0]
    half_mean_shape = int(mean2.shape[0]/2)

    mean_est1 = pd.DataFrame(mean2[0:half_mean_shape])
    Cov_est1 = pd.DataFrame(Cov2[0:half_mean_shape])

    mean_est2 = pd.DataFrame(mean2[half_mean_shape:mean_shape])
    Cov_est2 = pd.DataFrame(Cov2[half_mean_shape:mean_shape])
    
    #print(((Y[0:n_half_shape]-mean_est1)**2).mean(axis=ax))
    #print(((Y[n_half_shape:n_shape]-mean_est2)**2).mean(axis=ax))
    return ((Y-np.asarray(mean2))**2).mean(axis=0)[0]
    
   
```

```python
def plot_results(X_replicates, X, Y, m, model_number):
    num_predict = 100

    #X = data[0]
    #Y = data[1]

    n_shape = X.shape[0]
    n_half_shape = int(X.shape[0]/2)
    
    n_shape_replicates = X_replicates.shape[0]
    n_half_shape_replicates = int(X_replicates.shape[0]/2)

    # [Xi, Xj] = np.meshgrid(np.linspace(np.min(X[:,0]), np.max(X[:,0]), num_predict), np.linspace(np.min(X[:,1]), np.max(X[:,1]), num_predict))
    # X2 = X[:,0:1].copy()

    # # We need to augument our test space to be a list of coordinates for input to the GP
    # Xnew2 = np.vstack((Xi.ravel(), Xj.ravel())).T # Change our input grid to list of coordinates

    [exp1_mesh1, exp1_mesh2] = np.meshgrid(np.linspace(np.min(X[0:n_half_shape,0]), np.max(X[0:n_half_shape,0]), num_predict), np.linspace(np.min(X[0:n_half_shape,1]), np.max(X[0:n_half_shape,1]), num_predict))
    [exp2_mesh1, exp2_mesh2] = np.meshgrid(np.linspace(np.min(X[n_half_shape:n_shape,0]), np.max(X[n_half_shape:n_shape,0]), num_predict), np.linspace(np.min(X[n_half_shape:n_shape,1]), np.max(X[n_half_shape:n_shape,1]), num_predict))
    Xnew2 = np.concatenate([np.vstack((exp1_mesh1.ravel(), exp1_mesh2.ravel())).T, np.vstack((exp2_mesh1.ravel(), exp2_mesh2.ravel())).T], axis=0)

    # Predict the mean and covariance of the GP fit at the test locations
    mean2, Cov2 = m.predict_f(Xnew2)

    mean2 = np.asarray(mean2)
    Cov2 = np.asarray(Cov2)

    mean_shape = mean2.shape[0]
    half_mean_shape = int(mean2.shape[0]/2)

    mean_est1 = pd.DataFrame(mean2[0:half_mean_shape].reshape(exp1_mesh1.shape))
    Cov_est1 = pd.DataFrame(Cov2[0:half_mean_shape].reshape(exp1_mesh1.shape))

    mean_est2 = pd.DataFrame(mean2[half_mean_shape:mean_shape].reshape(exp2_mesh1.shape))
    Cov_est2 = pd.DataFrame(Cov2[half_mean_shape:mean_shape].reshape(exp2_mesh1.shape))

    # set up a figure twice as wide as it is tall
    #fig = plt.figure(figsize=plt.figaspect(0.5))
    fig = plt.figure(figsize=(25,10))

    Z1 = mean2[0:half_mean_shape].reshape(exp1_mesh1.shape)
    Z2 = mean2[half_mean_shape:mean_shape].reshape(exp2_mesh2.shape)

    #===============
    #  First subplot
    #===============
    # set up the axes for the first plot
    ax = fig.add_subplot(1, 2, 1, projection='3d')

    # plot a 3D surface like in the example mplot3d/surface3d_demo
    surf = ax.plot_surface(exp1_mesh1, exp1_mesh2, Z1, rstride=1, cstride=1, cmap=cm.viridis,
                           linewidth=0, antialiased=False)
    cbar = fig.colorbar(surf, shrink=0.5, aspect=10)
    for t in cbar.ax.get_yticklabels():
                 t.set_fontsize(18)
    cbar.set_label('f mean', fontsize=22)
    ax.scatter(X_replicates[0:n_half_shape_replicates,0], X_replicates[0:n_half_shape_replicates,1], Y[0:n_half_shape_replicates], color='k',marker="x", s=70, label="Data")

    ax.set_xticks(np.linspace(0.0,0.4,6))
    ax.set_yticks(np.linspace(0.0,0.25,6))
    
    plt.tick_params(axis='x', labelsize=20)
    plt.tick_params(axis='y', labelsize=20)
    plt.tick_params(axis='z', labelsize=20)

    ax.set_xlabel('x0 (mg/kg)', fontsize=25, labelpad=15)
    ax.set_ylabel('x1 (mg/kg)', fontsize=25, labelpad=15)
    ax.set_zlabel('Gene expression', fontsize=25, labelpad=15)
    #ax.set_zlabel('Expression', fontsize=20, labelpad=15)

    #ax.set_zlim(np.min(Y),np.max(Y))

    ax.set_title('Experiment 1', fontsize=25)

    #ax.view_init(15)
    ax.view_init(-140, 30)
    ax.view_init( 20, -120)
    plt.legend(loc="upper right", fontsize=25)


    #===============
    # Second subplot
    #===============
    # set up the axes for the second plot
    ax = fig.add_subplot(1, 2, 2, projection='3d')

    #ax.plot_surface(exp2_mesh1, exp2_mesh2, Z2, rstride=10, cstride=10, cmap=cm.viridis,
    #                       linewidth=0, antialiased=False)
    
    surf = ax.plot_surface(exp2_mesh1, exp2_mesh2, Z2, rstride=1, cstride=1, cmap=cm.viridis,
                           linewidth=0, antialiased=False)
    cbar = fig.colorbar(surf, shrink=0.5, aspect=10)
    cbar.set_label('f mean', fontsize=22)
    
    for t in cbar.ax.get_yticklabels():
                 t.set_fontsize(18)

    ax.scatter(X_replicates[n_half_shape_replicates:n_shape_replicates,0], X_replicates[n_half_shape_replicates:n_shape_replicates,1], Y[n_half_shape_replicates:n_shape_replicates], color='k',marker="x", s=70, label="Data")


    plt.tick_params(axis='x', labelsize=20)
    plt.tick_params(axis='y', labelsize=20)
    plt.tick_params(axis='z', labelsize=20)

    ax.set_xticks(np.linspace(0.0,0.4,6))
    ax.set_yticks(np.linspace(0.0,250,6))
    ax.set_xlabel('x0 (mg/kg)', fontsize=25, labelpad=15)
    ax.set_ylabel('x2 (mg/kg)', fontsize=25, labelpad=15)
    ax.set_zlabel('Gene expression', fontsize=25, labelpad=15)
    #ax.set_zlabel('Expression', fontsize=20, labelpad=15)

    #ax.set_zlim(np.min(Y),np.max(Y))

    ax.set_title('Experiment 2', fontsize=25)
    
    plt.tight_layout(rect=[0.2, 0.5, 0.1, 1.0])
    plt.tight_layout()
    plt.subplots_adjust(wspace=0.0, hspace=0.0)  

    ax.dist = 10
    #ax.view_init(15)
    ax.view_init(-140, 30)
    ax.view_init( 20, -120)
    plt.legend(loc="upper right", fontsize=25)
    
    plt.savefig('simulated_data_model_'+str(model_number)+'.png', bbox_inches = 'tight',
        pad_inches = 0)
    plt.show()
```

```python
def load_data1(data, gene_id, samples_exp1, samples_exp2):
    
    y_exp1 = data[data['Unnamed: 0'] == gene_id][samples_exp1].to_numpy().reshape(-1,1)
    y_exp2 = data[data['Unnamed: 0'] == gene_id][samples_exp2].to_numpy().reshape(-1,1)
    
    TU_IMI_exp1 = metadata[metadata['experiment']==1]['conc_IMI'].to_numpy().reshape(-1,1)
    TU_IMI_exp2 = metadata[metadata['experiment']==2]['conc_IMI'].to_numpy().reshape(-1,1)

    TU_CLO_exp1 = metadata[metadata['experiment']==1]['conc_CLO'].to_numpy().reshape(-1,1)
    TU_CLO_exp2 = metadata[metadata['experiment']==2]['conc_CLO'].to_numpy().reshape(-1,1)

    TU_CYPRO_exp1 = metadata[metadata['experiment']==1]['conc_CYPRO'].to_numpy().reshape(-1,1)
    TU_CYPRO_exp2 = metadata[metadata['experiment']==2]['conc_CYPRO'].to_numpy().reshape(-1,1)
    
    df_exp1 = pd.concat([pd.DataFrame(TU_IMI_exp1), pd.DataFrame(TU_CLO_exp1), pd.DataFrame(y_exp1)], axis=1)
    df_exp2 = pd.concat([pd.DataFrame(TU_IMI_exp2), pd.DataFrame(TU_CYPRO_exp2), pd.DataFrame(y_exp2)], axis=1)
    
    df_exp1.columns = ['conc_IMI', 'conc_CLO', 'y_exp1']
    df_exp2.columns = ['conc_IMI', 'conc_CYPRO', 'y_exp2']
    df_exp1['exp'] = 'exp1'
    df_exp2['exp'] = 'exp2'
    df_exp1.columns = ['', '', '', '']
    df_exp2.columns = ['', '', '', '']
    
    df = pd.concat([df_exp1, df_exp2], axis=0)
    df.columns = ['X1', 'X2', 'Y', 'exp']
    df['Y'] = (df['Y'] - np.min(df['Y']))/(np.max(df['Y'])-np.min(df['Y']))
    #df.columns = ['conc_drug1', 'conc_drug2', 'y']
    #df['conc_drug12'] = df['conc_drug1']*df['conc_drug2']
    #df_X = pd.concat([df['conc_drug1'], df['conc_drug2'], df['conc_drug12']], axis=1)
    #df_X.columns = ['','','']
    #X_full = df[[]].to_numpy()
    #Y_full = df['y'].to_numpy().reshape(-1,1)
    #Y_full = (Y_full - np.min(Y_full))/(np.max(Y_full)-np.min(Y_full))
    
    return df
    
    
```
