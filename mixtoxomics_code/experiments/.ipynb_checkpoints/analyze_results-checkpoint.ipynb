```python
def load_data1(data, gene_id, samples_exp1, samples_exp2):

    y_exp1 = data[data['Unnamed: 0'] == gene_id][samples_exp1].to_numpy().reshape(-1,1)
    y_exp2 = data[data['Unnamed: 0'] == gene_id][samples_exp2].to_numpy().reshape(-1,1)

    TU_IMI_exp1 = metadata[metadata['experiment']==1]['conc_IMI'].to_numpy().reshape(-1,1)
    TU_IMI_exp2 = metadata[metadata['experiment']==2]['conc_IMI'].to_numpy().reshape(-1,1)

    TU_CLO_exp1 = metadata[metadata['experiment']==1]['conc_CLO'].to_numpy().reshape(-1,1)
    TU_CLO_exp2 = metadata[metadata['experiment']==2]['conc_CLO'].to_numpy().reshape(-1,1)

    TU_CYPRO_exp1 = metadata[metadata['experiment']==1]['conc_CYPRO'].to_numpy().reshape(-1,1)
    TU_CYPRO_exp2 = metadata[metadata['experiment']==2]['conc_CYPRO'].to_numpy().reshape(-1,1)

    df_exp1 = pd.concat([pd.DataFrame(TU_IMI_exp1), pd.DataFrame(TU_CLO_exp1), pd.DataFrame(y_exp1)], axis=1)
    df_exp2 = pd.concat([pd.DataFrame(TU_IMI_exp2), pd.DataFrame(TU_CYPRO_exp2), pd.DataFrame(y_exp2)], axis=1)

    df_exp1.columns = ['conc_IMI', 'conc_CLO', 'y_exp1']
    df_exp2.columns = ['conc_IMI', 'conc_CYPRO', 'y_exp2']
    df_exp1['exp'] = 'exp1'
    df_exp2['exp'] = 'exp2'
    df_exp1.columns = ['', '', '', '']
    df_exp2.columns = ['', '', '', '']

    df = pd.concat([df_exp1, df_exp2], axis=0)
    df.columns = ['X1', 'X2', 'Y', 'exp']
    df['Y'] = (df['Y'] - np.min(df['Y']))/(np.max(df['Y'])-np.min(df['Y']))
    #df.columns = ['conc_drug1', 'conc_drug2', 'y']
    #df['conc_drug12'] = df['conc_drug1']*df['conc_drug2']
    #df_X = pd.concat([df['conc_drug1'], df['conc_drug2'], df['conc_drug12']], axis=1)
    #df_X.columns = ['','','']
    #X_full = df[[]].to_numpy()
    #Y_full = df['y'].to_numpy().reshape(-1,1)
    #Y_full = (Y_full - np.min(Y_full))/(np.max(Y_full)-np.min(Y_full))

    return df
```

```python
#for i in range(300,400):
# Load the data to extract input points
gene_id = genes[i]
gene_id = 'Fcan01_00004' # model 3
#gene_id = 'Fcan01_00009'# model 2

data = load_data1(data_full, gene_id, samples_exp1, samples_exp2)
data = (data[['X1', 'X2']].to_numpy(), data['Y'].to_numpy().reshape(-1,1))

## Pick which model to fit 
best_model = 3

# model 1
if(best_model == 1):
    k = K_full_model_1()

    m = gpflow.models.GPR(data, kernel=k, mean_function=None)
    m.likelihood.variance.assign(0.001)

    opt = gpflow.optimizers.Scipy()
    opt_logs = opt.minimize(m.training_loss, m.trainable_variables, options=dict(maxiter=100))
    
    loglik_model_1 = np.asarray(m.training_loss())

    loglik_model_1_list.append(loglik_model_1)

# model 2
if(best_model == 2):

    k = K_full_model_2()
    m = gpflow.models.GPR(data, kernel=k, mean_function=None)
    m.likelihood.variance.assign(0.001)

    opt = gpflow.optimizers.Scipy()
    opt_logs = opt.minimize(m.training_loss, m.trainable_variables, options=dict(maxiter=100))
    
    loglik_model_2 = np.asarray(m.training_loss())

    loglik_model_2_list.append(loglik_model_2)

# model 3
if(best_model == 3):
    k = K_full_model_3()
    m = gpflow.models.GPR(data, kernel=k, mean_function=None)
    m.likelihood.variance.assign(0.001)

    opt = gpflow.optimizers.Scipy()
    opt_logs = opt.minimize(m.training_loss, m.trainable_variables, options=dict(maxiter=100))
    
    loglik_model_3 = np.asarray(m.training_loss())

    loglik_model_3_list.append(loglik_model_3)

# model 4
if(best_model == 4):
    k = K_full_model_4()
    m = gpflow.models.GPR(data, kernel=k, mean_function=None)
    m.likelihood.variance.assign(0.001)

    opt = gpflow.optimizers.Scipy()
    opt_logs = opt.minimize(m.training_loss, m.trainable_variables, options=dict(maxiter=100))
    
    loglik_model_4 = np.asarray(m.training_loss())

    loglik_model_4_list.append(loglik_model_4)

# model 5
if(best_model == 5):
    k = K_full_model_5()
    m = gpflow.models.GPR(data, kernel=k, mean_function=None)
    m.likelihood.variance.assign(0.001)

    opt = gpflow.optimizers.Scipy()
    opt_logs = opt.minimize(m.training_loss, m.trainable_variables, options=dict(maxiter=100))
    
    loglik_model_5 = np.asarray(m.training_loss())

    loglik_model_5_list.append(loglik_model_5)


# model 6
if(best_model == 6):
    k = K_full_model_6()
    m = gpflow.models.GPR(data, kernel=k, mean_function=None)
    m.likelihood.variance.assign(0.001)

    opt = gpflow.optimizers.Scipy()
    opt_logs = opt.minimize(m.training_loss, m.trainable_variables, options=dict(maxiter=100))
    
    loglik_model_6 = np.asarray(m.training_loss())

    loglik_model_6_list.append(loglik_model_6)


# model 7
if(best_model == 7):
    k = K_full_model_7()
    m = gpflow.models.GPR(data, kernel=k, mean_function=None)
    m.likelihood.variance.assign(0.001)

    opt = gpflow.optimizers.Scipy()
    opt_logs = opt.minimize(m.training_loss, m.trainable_variables, options=dict(maxiter=100))
    
    loglik_model_7 = np.asarray(m.training_loss())

    loglik_model_7_list.append(loglik_model_7)

# model 8
if(best_model == 8):
    k = K_full_model_8()
    m = gpflow.models.GPR(data, kernel=k, mean_function=None)
    m.likelihood.variance.assign(0.001)

    opt = gpflow.optimizers.Scipy()
    opt_logs = opt.minimize(m.training_loss, m.trainable_variables, options=dict(maxiter=100))
    
    loglik_model_8 = np.asarray(m.training_loss())

    loglik_model_8_list.append(loglik_model_8)

# model 9
if(best_model == 9):
    k = K_full_model_9()
    m = gpflow.models.GPR(data, kernel=k, mean_function=None)
    m.likelihood.variance.assign(0.001)

    opt = gpflow.optimizers.Scipy()
    opt_logs = opt.minimize(m.training_loss, m.trainable_variables, options=dict(maxiter=100))
    
    loglik_model_9 = np.asarray(m.training_loss())

    loglik_model_9_list.append(loglik_model_9)

# model 10
if(best_model == 10):
    k = K_full_model_10()
    m = gpflow.models.GPR(data, kernel=k, mean_function=None)
    m.likelihood.variance.assign(0.001)

    opt = gpflow.optimizers.Scipy()
    opt_logs = opt.minimize(m.training_loss, m.trainable_variables, options=dict(maxiter=100))   
    
    loglik_model_10 = np.asarray(m.training_loss())

    loglik_model_10_list.append(loglik_model_10)

# model 11
if(best_model == 11):
    k = K_full_model_11()
    m = gpflow.models.GPR(data, kernel=k, mean_function=None)
    m.likelihood.variance.assign(0.001)

    opt = gpflow.optimizers.Scipy()
    opt_logs = opt.minimize(m.training_loss, m.trainable_variables, options=dict(maxiter=100))
    
    loglik_model_11 = np.asarray(m.training_loss())

    loglik_model_11_list.append(loglik_model_11)

print_summary(m)

num_predict = 100

X = data[0]
Y = data[1]

n_shape = X.shape[0]
n_half_shape = int(X.shape[0]/2)

# [Xi, Xj] = np.meshgrid(np.linspace(np.min(X[:,0]), np.max(X[:,0]), num_predict), np.linspace(np.min(X[:,1]), np.max(X[:,1]), num_predict))
# X2 = X[:,0:1].copy()

# # We need to augument our test space to be a list of coordinates for input to the GP
# Xnew2 = np.vstack((Xi.ravel(), Xj.ravel())).T # Change our input grid to list of coordinates

[exp1_mesh1, exp1_mesh2] = np.meshgrid(np.linspace(np.min(X[0:n_half_shape,0]), np.max(X[0:n_half_shape,0]), num_predict), np.linspace(np.min(X[0:n_half_shape,1]), np.max(X[0:n_half_shape,1]), num_predict))
[exp2_mesh1, exp2_mesh2] = np.meshgrid(np.linspace(np.min(X[n_half_shape:n_shape,0]), np.max(X[n_half_shape:n_shape,0]), num_predict), np.linspace(np.min(X[n_half_shape:n_shape,1]), np.max(X[n_half_shape:n_shape,1]), num_predict))
Xnew2 = np.concatenate([np.vstack((exp1_mesh1.ravel(), exp1_mesh2.ravel())).T, np.vstack((exp2_mesh1.ravel(), exp2_mesh2.ravel())).T], axis=0)

# Predict the mean and covariance of the GP fit at the test locations
mean2, Cov2 = m.predict_f(Xnew2)

mean2 = np.asarray(mean2)
Cov2 = np.asarray(Cov2)

mean_shape = mean2.shape[0]
half_mean_shape = int(mean2.shape[0]/2)

mean_est1 = pd.DataFrame(mean2[0:half_mean_shape].reshape(exp1_mesh1.shape))
Cov_est1 = pd.DataFrame(Cov2[0:half_mean_shape].reshape(exp1_mesh1.shape))

mean_est2 = pd.DataFrame(mean2[half_mean_shape:mean_shape].reshape(exp2_mesh1.shape))
Cov_est2 = pd.DataFrame(Cov2[half_mean_shape:mean_shape].reshape(exp2_mesh1.shape))

# set up a figure twice as wide as it is tall
#fig = plt.figure(figsize=plt.figaspect(0.5))
fig = plt.figure(figsize=(25,10))

Z1 = mean2[0:half_mean_shape].reshape(exp1_mesh1.shape)
Z2 = mean2[half_mean_shape:mean_shape].reshape(exp2_mesh2.shape)

Cov_est1 = np.asarray(Cov_est1)
Cov_est2 = np.asarray(Cov_est2)

Z1_cov = Cov_est1.reshape(exp1_mesh1.shape)
Z2_cov = Cov_est2.reshape(exp2_mesh2.shape)

Z2_95_l_exp2 = Z2 - 1.96*np.sqrt(Z2_cov)
Z2_95_u_exp2 = Z2 + 1.96*np.sqrt(Z2_cov)

Z1_95_l_exp1 = Z1 - 1.96*np.sqrt(Z1_cov)
Z1_95_u_exp1 = Z1 + 1.96*np.sqrt(Z1_cov)

#===============
#  First subplot
#===============
# set up the axes for the first plot
ax = fig.add_subplot(1, 2, 1, projection='3d')

# plot a 3D surface like in the example mplot3d/surface3d_demo
surf = ax.plot_surface(exp1_mesh1, exp1_mesh2, Z1, rstride=1, cstride=1, cmap=cm.viridis,
                       linewidth=0, antialiased=False)

ax.plot_surface(exp1_mesh1, exp1_mesh2, Z1_95_l_exp1, rstride=10, cstride=10, cmap='gray',
                       linewidth=0, antialiased=False, alpha=0.3)

ax.plot_surface(exp1_mesh1, exp1_mesh2, Z1_95_u_exp1, rstride=10, cstride=10, cmap='gray',
                       linewidth=0, antialiased=False, alpha=0.3)


cbar = fig.colorbar(surf, shrink=0.5, aspect=10)
for t in cbar.ax.get_yticklabels():
             t.set_fontsize(18)

cbar.set_label('Scaled Expression', fontsize=22)
ax.scatter(X[0:n_half_shape,0], X[0:n_half_shape,1], Y[0:n_half_shape])

ax.set_xticks(np.linspace(0.0,0.4,6))
ax.set_yticks(np.linspace(0.0,0.25,6))
plt.tick_params(axis='x', labelsize=20)
plt.tick_params(axis='y', labelsize=20)
plt.tick_params(axis='z', labelsize=20)

ax.set_xlabel('IMI (mg/kg)', fontsize=25, labelpad=15)
ax.set_ylabel('CLO (mg/kg)', fontsize=25, labelpad=15)
#ax.set_zlabel('Scaled Expression', fontsize=22, labelpad=15)

#ax.set_zlim(np.min(Y),np.max(Y))

ax.set_title('Experiment 1  ' + '('+str(gene_id)+ ')', fontsize=25)



plt.subplots_adjust(wspace=0.0, hspace=0.01)  

#ax.view_init(15)
ax.view_init(-140, 30)
ax.view_init( 20, -120)


#===============
# Second subplot
#===============
# set up the axes for the second plot
ax = fig.add_subplot(1, 2, 2, projection='3d')

surf = ax.plot_surface(exp2_mesh1, exp2_mesh2, Z2, rstride=10, cstride=10, cmap=cm.viridis,
                       linewidth=0, antialiased=False)

ax.plot_surface(exp2_mesh1, exp2_mesh2, Z2_95_l_exp2, rstride=10, cstride=10, cmap='gray',
                       linewidth=0, antialiased=False, alpha=0.3)

ax.plot_surface(exp2_mesh1, exp2_mesh2, Z2_95_u_exp2, rstride=10, cstride=10, cmap='gray',
                       linewidth=0, antialiased=False, alpha=0.3)


cbar = fig.colorbar(surf, shrink=0.5, aspect=10)
for t in cbar.ax.get_yticklabels():
             t.set_fontsize(18)

cbar.set_label('Scaled Expression', fontsize=22)
        
ax.scatter(X[n_half_shape:n_shape,0], X[n_half_shape:n_shape,1], Y[n_half_shape:n_shape])


plt.tick_params(axis='x', labelsize=20)
plt.tick_params(axis='y', labelsize=20)
plt.tick_params(axis='z', labelsize=20)

ax.set_xticks(np.linspace(0.0,0.4,6))
ax.set_yticks(np.linspace(0.0,250,6))
ax.set_xlabel('IMI (mg/kg)', fontsize=25, labelpad=15)
ax.set_ylabel('CYPRO (mg/kg)', fontsize=25, labelpad=15)
#ax.set_zlabel('Scaled Expression', fontsize=22, labelpad=15)

#ax.set_zlim(np.min(Y),np.max(Y))

ax.set_title('Experiment 2  ' + '('+str(gene_id)+ ')', fontsize=25)

ax.dist = 10
#ax.view_init(15)
ax.view_init(-140, 30)
ax.view_init( 20, -120)

plt.tight_layout(rect=[0.2, 0.5, 0.1, 1.0])

plt.tight_layout()

plt.subplots_adjust(wspace=0.0, hspace=0.0)  
#fig.subplots_adjust(left=0.2)

plt.savefig(str(gene_id)+'_model_'+str(best_model)+'_3d.png', bbox_inches = 'tight',
    pad_inches = 0)
plt.show()
```
