```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import gpflow
from matplotlib import cm
from matplotlib.ticker import LinearLocator, FormatStrFormatter
from matplotlib.ticker import LinearLocator, FormatStrFormatter
f64 = gpflow.utilities.to_default_float
np.set_printoptions(suppress=True)
```

```python
# Load custom code for HandGP model
# See also repository https://github.com/YuliyaShapovalova/HandGP
from utilities import (predict_in_observations_lower, predict_in_observations_upper, compute_prior_hyperparameters, trapezoidal_area, predict_in_observations, fit_Hand, K_multiplicative)
```

### Load the data

```python
## Imi-Clo experiment
imi_clo_reproduction = pd.read_excel("../data/reproduction/Counts_reproduction_IMI_CLO.xlsx")
```

```python
## Imi-Cypro experiment
imi_cypro_reproduction = pd.read_excel("../data/reproduction/Counts_reproduction_IMI_CYPRO.xlsx")
```

```python
## Normalize and create right data format Imi-Clo experiment
X_imi_clo = np.concatenate([imi_clo_reproduction['conc IMI'].to_numpy().reshape(-1,1), imi_clo_reproduction['conc CLO'].to_numpy().reshape(-1,1)], axis=1)
Y_imi_clo = imi_clo_reproduction['Juveniles'].to_numpy().reshape(-1,1)
Y_imi_clo = (Y_imi_clo - np.min(Y_imi_clo))/(np.max(Y_imi_clo)-np.min(Y_imi_clo))
```

```python
## Normalize and create right data format Imi-Cypro experiment
X_imi_cypro = np.concatenate([imi_cypro_reproduction['concentration (mg/kg) Imidacloprid'].to_numpy().reshape(-1,1), imi_cypro_reproduction['concentration (mg/kg) Cyproconazole'].to_numpy().reshape(-1,1)], axis=1)
Y_imi_cypro = imi_cypro_reproduction['juveniles'].to_numpy().reshape(-1,1)
Y_imi_cypro = (Y_imi_cypro-np.min(Y_imi_cypro))/(np.max(Y_imi_cypro)-np.min(Y_imi_cypro))
```

```python
# Rename data frame columns for Imi-Clo experiment
imi_clo_reproduction.columns = ['Rankings', 'Concentrations', 'ToxicUnitsIMI', 'doseIMI', 'ToxicUnitsCLO', 'doseCLO', 'series', 'drug1.conc', 'drug2.conc', 'effect', 'Adults', 'Remarks']
```

```python
# Rename data frame columns for Imi-Cypro experiment
imi_cypro_reproduction.columns = ['ToxicUnits Imidacloprid', 'ToxicUnits Cyproconazole', 'concentration (mg/kg) Imidacloprid', 'TU imi reverse calculated',  'drug2.conc','TU cypro reverse calculated','drug1.conc', 'adults', 'effect']
```

## Build and fit two GP models (m1 for Imi-Clo experiemnt, m2 for Imi-Cypro experiemnt)

```python
# The concentrations X_imi_clo and responses Y_imi_clo are combined to feed into the model
data_imi_clo = (X_imi_clo, Y_imi_clo)

# The kernel is defined
k = K_multiplicative()

# The model is defined
m1 = gpflow.models.GPR(data_imi_clo, kernel=k, mean_function=None)

# Optimization of the model
opt = gpflow.optimizers.Scipy()
opt_logs = opt.minimize(m1.training_loss, m1.trainable_variables, options=dict(maxiter=100))
```

```python
# The concentrations X_imi_clo and responses Y_imi_clo are combined to feed into the model
data_imi_cypro = (X_imi_cypro, Y_imi_cypro)

# The kernel is defined
k = K_multiplicative()

# The model is defined
m2 = gpflow.models.GPR(data_imi_cypro, kernel=k, mean_function=None)

# Optimization of the model
opt = gpflow.optimizers.Scipy()
opt_logs = opt.minimize(m2.training_loss, m2.trainable_variables, options=dict(maxiter=100))
```

### Fit of the HandGP model for Imi-Clo experiemnt

```python
## Predict the mean and covariance of the GP fit at the test locations

# Creating grid of points where we make predictions
num_predict = 50
[exp1_mesh1, exp1_mesh2] = np.meshgrid(np.linspace(np.min(data_imi_clo[0][:,0]), np.max(data_imi_clo[0][:,0]), num_predict), np.linspace(np.min(data_imi_clo[0][:,1]), np.max(data_imi_clo[0][:,1]), num_predict))
Xnew2 = np.vstack((exp1_mesh1.ravel(), exp1_mesh2.ravel())).T

# These are predictions of the GP mean (mean2) and covariance (Cov2) for Imi-Clo experiement
mean2, Cov2 = m1.predict_f(Xnew2)
mean2 = np.asarray(mean2)
Cov2 = np.asarray(Cov2)
```

```python
# The Concentrations and Effects are separated into individual ones for the HandGP model fits

Effect = imi_clo_reproduction['effect'].values.reshape(-1,1).copy()
Dose_A = imi_clo_reproduction['drug1.conc'].values.astype(float).copy()
Dose_B = imi_clo_reproduction['drug2.conc'].values.astype(float).copy()

Dose_AB = np.concatenate((Dose_A.reshape(-1,1), Dose_B.reshape(-1,1)),  axis=1)

Effect_B = imi_clo_reproduction[imi_clo_reproduction['drug1.conc'] == 0]['effect'].to_numpy().reshape(-1,1).astype(float)
Effect_A = imi_clo_reproduction[imi_clo_reproduction['drug2.conc'] == 0]['effect'].to_numpy().reshape(-1,1).astype(float)
Dose_A = imi_clo_reproduction[imi_clo_reproduction['drug2.conc']==0]['drug1.conc'].to_numpy().reshape(-1,1).astype(float)
Dose_B = imi_clo_reproduction[imi_clo_reproduction['drug1.conc']==0]['drug2.conc'].to_numpy().reshape(-1,1).astype(float)

eff_max_a = np.max(Effect_A)
eff_max_b = np.max(Effect_B)

eff_min_a = np.min(Effect_A)
eff_min_b = np.min(Effect_B)
```

```python
# The mean and covariance of the GP as pandas data frame

mean2 = pd.DataFrame(np.asarray(mean2).reshape(exp1_mesh1.shape))
Cov2 = pd.DataFrame(np.asarray(Cov2).reshape(exp1_mesh1.shape))
```

```python
## HandGP model fit for Imi-Clo experiment

# First individual GPs are isolated
dim2_A = mean2.iloc[0].to_numpy()
dim2_B = mean2.loc[:][0].to_numpy()

dim2_A_lower = mean2.loc[0].to_numpy() - 1.96 * np.sqrt(Cov2.loc[0]).to_numpy()
dim2_B_lower = mean2.iloc[:,0].to_numpy() - 1.96 * np.sqrt(Cov2.iloc[:,0]).to_numpy()

dim2_A_upper = mean2.loc[0].to_numpy() + 1.96 * np.sqrt(Cov2.loc[0]).to_numpy()
dim2_B_upper = mean2.iloc[:,0].to_numpy() + 1.96 * np.sqrt(Cov2.iloc[:,0]).to_numpy()

data = pd.concat([pd.DataFrame(Dose_AB), pd.DataFrame(Effect)], axis=1)
data.columns = ['Dose_A', 'Dose_B', 'Effect']

X1 = data[data['Dose_B']==0]['Dose_A'].to_numpy().reshape(-1,1).astype(float)
Y1 = data[data['Dose_B']==0]['Effect'].to_numpy().reshape(-1,1).astype(float)

X2 = data[data['Dose_A']==0]['Dose_B'].to_numpy().reshape(-1,1).astype(float)
Y2 = data[data['Dose_A']==0]['Effect'].to_numpy().reshape(-1,1).astype(float)

# Expected HandGP surfaces are computed (Y_expected_Hand -- mean, Y_expected_Hand_lower 95% lower bound, 95% Y_expected_Hand_upper bound)
Y_expected_Hand = fit_Hand(X1, X2, dim2_A, dim2_B, Dose_A, Dose_B)

Y_expected_Hand_lower = fit_Hand(X1, X2, dim2_A_lower, dim2_B_lower, Dose_A, Dose_B)

Y_expected_Hand_upper = fit_Hand(X1, X2, dim2_A_upper, dim2_B_upper, Dose_A, Dose_B)
```

```python
# Preidctions for the GP mean, lower and upper bounds

mean_full, Cov_full = predict_in_observations(X1, X2, m1)

mean_full_lower, Cov_full_lower = predict_in_observations_lower(X1, X2, m1)

mean_full_upper, Cov_full_upper = predict_in_observations_upper(X1, X2, m1)
```

### Plot the results for Imi-Clo experiement

#### Here we plot the difference between expected surface according to HandGP model and the surface based only on observations

```python
fig, (ax1, ax2, ax3) = plt.subplots(1, 3, figsize=(16,5), gridspec_kw={'width_ratios': [1, 1, 1.25]})
v = np.linspace(-0.4, 0.4, 8, endpoint=True)
cf1 = ax1.contourf(Dose_A.flatten(), Dose_B.flatten(),Y_expected_Hand_lower - mean_full_lower,v,vmin=-0.4, vmax=0.4,cmap='RdYlGn')
ax1.tick_params(axis='x', labelsize=12)
ax1.tick_params(axis='y', labelsize=12)
ax1.set_xlabel('Imidacloprid (mg/kg)', fontsize=16)
ax1.set_ylabel('Clothianidin (mg/kg)', fontsize=16)
ax1.set_title('95% lower bound CI', fontsize=16)

cf2 = ax2.contourf(Dose_A.flatten(), Dose_B.flatten(),Y_expected_Hand - mean_full,v,vmin=-0.4, vmax=0.4,cmap='RdYlGn')

ax2.tick_params(axis='x', labelsize=12)
ax2.tick_params(axis='y', labelsize=12)
ax2.set_title('Predicted Effect', fontsize=16)

cf3 = ax3.contourf(Dose_A.flatten(), Dose_B.flatten(),Y_expected_Hand_upper - mean_full_upper, v,cmap='RdYlGn')

ax3.tick_params(axis='x', labelsize=12)
ax3.tick_params(axis='y', labelsize=12)
ax3.set_title('95% upper bound CI', fontsize=16)
ticks = np.linspace(-0.4, 0.4, 5)
cbar = fig.colorbar(cf3, ax=ax3, shrink=0.9,  ticks=ticks)
cbar.set_label('Antagonism | Synergy ', rotation=270, fontsize=16, labelpad=15)
plt.savefig('imi_clo.pdf', pad_inches = 1)
plt.show()
```
